```python
import re
```

```python
text_to_search = """
abcdefghijklmnopqrstuvwxyz
ABCDEFGHIJKLMNOPQRSTUVWXYZ
1234567890

Ha HaHa

MetaCharacters (Need to be escaped):
. ^ $ + ? { } [ ] \\ | ( )

coreyms.com

321-555-4321
123.555.1234
123*555*1234
800-555-1267
900-555-1289

Mr. Schafter
Mr Smith
Ms Davis
Mrs. Robinson
Mr. T

cat
mat
pat
bat
"""

sentence = "Start a sentence and then bring it to the end"
```

**Raw string** - This is a string prefixed with **r** in python eg r"\tTab"

In python, the \t is interpreted as tab but with the r at the begining, it is a raw string which is to be interpreted by regex.

To write our patterns, we are going to use the re.compile method which will allow us to seperate out our methods into a variable and make that variable reusable to perform multiple searches.

```python
#Let's say we wanted to search for literal characters "abc"
pattern = re.compile(r"abc")

#We are going to use finditer which returns an iterator which we can loop through to get more infor through
#attributes like .group(), .span(), .start(), .end()
#It returns an iterator that contains all the matches
matches = pattern.finditer(text_to_search)

for match in matches:
    print(f"Match found:\t{match.group()}")
    print(f"Positions:\t{match.span()}")
    print(f"Start:\t{match.start()}")
    print(f"End:\t{match.end()}")



# The indexes - span are useful because they allow us to use the sting slicing functionality in python, 
# where we could just plug in these values and get the exact match

#Let's print out our string from index 28 to 31
# print(text_to_search[1:4])
```

```python
#Escaping meta characters
#Lets say for example we wanted to escape the dot in coreyms.com, we use the backward slash

pattern2 = re.compile(r"coreyms\.com")

matches2 = pattern2.finditer(text_to_search)

for match2 in matches2:
    print(f"Match found:\t{match2.group()}")
    print(f"Positions:\t{match2.span()}")
    print(f"Start:\t{match2.start()}")
    print(f"End:\t{match2.end()}")
```

Characters we can match:

- .  : Any character Except New line
- \d : Digits (0-9)
- \D : Not a digit
- \w : Word character (a-z, A-Z, 0-9, _)
- \W : Not a word character
- \s : Whitespace (space, tab, newline)
- \S : Not whitespace

Anchors:
The dont match any characters rather they match invisible positions before or after characters 

- \b - Word boundary
word bondaries are indicated by white spaces or a non alpha numeric character.
- \B - Not a word boundary
- ^  - Begining of the string
- $ - End of the string

Quantifiers

It allows us to specify the amount of digits we are looking for with out having to type them all out as we have seen in the previous examples

They include:

* \* - 0 or more
* \+ - 1 or more
* ? - 0 or one
* {3} - Exact number
* {3,4} - Range of number(min, max)

```python
# Let's say we want to match this names from our text to search: 
# Mr. Schafter
# Mr. Smith
# Ms Davis
# Mrs. Robinson
# Mr. T
pattern26 = re.compile(r"M[rs]s?\W+\w+")

matches26 = pattern26.finditer(text_to_search)

for match26 in matches26:
    print(match26.group())
```

Capturing information from groups

```python
urls = '''
https://www.google.com
http://coreyms.com
https://youtube.com
https://www.nasa.gov
'''
```

```python
# Trial
# Let's say we want to capture the domain name and the top level domain
# To capture this sections we can just put them in groups by surrounding them with parenthesis

pat3 = re.compile(r"https?://[a-zA-Z._-]+")
pat3.findall(urls)

#This one is best for messy real world data 
```

```python
# Let's say we want to capture the domain name and the top level domain
# To capture this sections we can just put them in groups by surrounding them with parenthesis

pattern30 = re.compile(r"https?://(www\.)?(\w+)(\.\w+)")
#Now we have three groups, the optional "www.", the domain name and the top level domain
#Group zero is everything that we captured-the whole url

urls_matched = pattern30.finditer(urls)

count_ = 1
for url in urls_matched:
    
    print("-"*50, f"url:{count_}")
    print(f"Group 0:\t{url.group(0)}\nGroup1:\t{url.group(1)}\nGroup2:\t{url.group(2)}\nGroup3:\t{url.group(3)}")
    
    count_+= 1

#This is the textbook version
```

Back refference: This is a way to refer to parts of a match either within the match itself or after we have mathced it or when we are doing a string replacement. 

```python
#Let's say for example we want to match characters with single quotes or double quotes at the 
# begining and the same at the end

pat = re.compile(r"('|\")\w+\"") 

print(pat.findall("food"))
```

```python
pattt = re.compile(r"(https?://)(www.)?([a-zA-Z0-9._-]+)")
pattss = pattt.finditer(urls)
for patss in pattss:
    print(patss.group())
```

```python
subs = pattt.sub(r"\3", urls)
print(subs)
```

```python
#Let's now get the domain name and top level domain
#We can use back reference to refference our captured group
# It is basiclly a short cut for accessing these groups using group indexes

#We will use the sub method to perform the substitution
#We will substitute the backrefferences which reference the groups

# in the sub method, the substitution that we are going to use are the back refferences that 
# reference the groups that we want 
#We want to replace the urls with the domain name(\2) and top level domain(\3) 



subed_url = pattern30.sub(r"\2\3", urls)
print(subed_url)

#What this does is that, every time it finds a match - pattern30, it replaces it with the domain 
# name and the top level damain as we have backrefferenced which we have back refferenced
```

Other methods we can use apart from finditer:

* re.sub() - for substitution as we have seen in the above example
* re.findall() - it works the same as finditer only that it returns the matches as a list of strings while finditer returns match objects with extra information and functionality
If it is matching groups, it will only return groups
If there are multiple groups, it will return a tuple of groups
If there are no groups it will just return a list of all the matches.

* re.match() - It will determine if the regular expression matches at the begining of the string

Example:

```python
print(sentence)
#We have variable named sentence and we want to match the first worrd in the sentence

pattern31=re.compile(r"Start")
match31 = pattern31.match(sentence)
print(match31.group())

# If we try to match anything that is in the sentence but not at the begining of the sentence, we 
# get none
# Example
print("-"*50)
pattern32=re.compile(r"bring")
match32= pattern32.match(sentence)
print(match32)

# It only matches things at the begining of strings
```

re.search() - If we wanna search for matches within the entire string, we use the search method
This only finds the first occurence of the match
Example

```python
#Let's look for bring, this time using the search method instead of the match method.

match33 = pattern32.search(sentence)
print(match33) 

# It only prints out the first match that it finds
# If we search for something that doesn't match it returns none
# Example
print("-"*50)
pattern33 = re.compile("kithusi")
match34=pattern33.search(sentence)
print(match34)
```

##### Flags

We can use flags like "flags=re.I"- which matches searches irregardless of there case

Flags make our work easier

Example:

```python
#Lets say we wanted to match a word but match it whether in uppercase or lowercase, match both
#Let's search for BRING

pattern34 = re.compile("BRING", re.I)
match35=pattern34.search(sentence)
print(match35)

#The "bring" in our sentence is lower case but in our expression we have uppercase.
#As we know regex is case sensitive but with the help of the flag "re.I", we can anything 
#regardless of there case
```

```python
# Let's extract everything at once but in groups, date, time and location

text2 = "2023/07/14 11:24PM Nairobi"

#This is a universal pattern incase we are doing data cleaning
pattern5 = re.compile(r"(\d+[/-]\d+[/-]\d+)\s?(\d{2}:\d{2}[AP]M)\s?(\w+)")

#Otherwise, if we know what we are looking for and the fomart is consistent
# pattern5 = re.compile(r"(\d{4}/\d{2}/\d{2})\s(\d{2}:\d{2}[AP]M)\s(\w+)")

match5=pattern5.finditer(text2)

for match_ in match5:
    print(f"Date:\t{match_.group(1)}")
    print(f"Time:\t{match_.group(2)}")
    print(f"Location:\t{match_.group(3)}")
```

## **Advanced string manipulation - Regex**
---
A regular expression, commonly known as a **regex**, is a sequence of characters used to define a search pattern.

It's a powerful tool in python for handling strings.

Why use use regex:

- It is essential for finding or matching patterns in a text, a common requiremnet in data processing
- It enables complex text transformation and processing with minimal code, making tasks more efficient and less error prone.
- It is widely used for parsing text data and extracting information like emails, phone numbers, or specific formats
- Python regex can handle large volumes of data, making it suitable for big data processing and automation.

#### **Literal characters**
---
Literal characters are the simplest form of matching.

In regex, most characters, including all letters and numbers are literal characters.

**Basics**

When we use a literal character in a regex pattern, the engine looks for an exact match of that character in the search string.

For exaple, the regex 'a' will match the character 'a' in the string 'cat', but not 'c', or 't'

**Non-meta characters**

Apart from a few special characters(like .*? etc), most characters are considered literal in regular expression.

For example, characters like b,x,5, or = are literal characters.

**Case sensitive**

Regular expressions are case sensitive by default.

This means that 'a' and 'A' are considered different characters.

For example, the regex 'cat' will match 'cat' but not 'Cat' or 'CAT'.

**Combining with meta characters**

Literal characters can be combined with meta characters to form more complex patterns.

For example, ca+t will match with 'cat', 'caat', 'caaat', etc, where 'a' is a literal character and + is a meta character.

#### Quantifiers and repetitions
---
Quantifiers and repetitions are fundamental components of regex that allow us to specify how many times a particular pattern should be matched.

**Basic quantifiers**

---

* __\*(Asterisk)__: Matches the preceding element, zero or more times.

For example, 'ab*' will match 'a', 'ab', 'abb', 'abbb', etc.

* __+(plus)__ : Matches the preceding element one or more times.

For example, 'ab+' will match 'ab', 'abb', 'abbb', etc, but not 'a'.

* __?(Question mark)__ : Makes the preceding element optional, matching either once or not at all.

For example, a? will match 'a', or nothing.

**Specific quantifiers**

---

* __{} (Curley)__ : Used to specify an exact number or range of repetitions:

* * {n} (Exact number) : The preceding element is matched exactly n times.

We use them to control exactly how many times a pattern is repeated.

For example, a{3} will match aaa only.

* * __{n,m} (Range)__ : Matches the preceeding element atleast n times but not more than m times.

For example, a{2,4} will match 'aa', 'aaa', and 'aaaa'.

* * __{n,} (Atleast n)__ : Matches the preceding element atleast n times.

For example, a{3,} matches 'aaa', 'aaaa', 'aaaaaaa', etc but not 'aa' or 'a'

**Greedy vs Lazy matching** 

---
Regular quantifires like *, +, {n,} can behave in two ways:

___
1. **Greedy Matching** (Default): Meaning match as much as possible.
___

It tries to take the longest possible string that still fits the pattern.

Regular expressions are by default greedy, meaning they match as many occurences as possible.

For example in 'aabbbb', 'ab*' will match 'abbbb'

Greedy quantifiers attempt to match as much text as possible given an entire regex pattern.

HTML code is a classic example to demonstrate this.


```python
import re
```

```python
greedy = re.compile("<.*>")
print(greedy.findall("<title>Regex</title>"))
```

If we look at the code, we may expect that matching is going to result in the first part of the string- \<title> and the second part of the string - \<\title\> because we are using find all, however, we will only get one entire string - ['\<title>Regex\</title>']

The reason for this is that the asterisk is greedy so it is going to much as many characters as possible given the entire string and entitre pattern.

Notice that even though the text has two pairs of tags <.....>, the matching only resulted in one match.

This is the behavior of greedy qyantifiers.

The "greedy" means the __*__ metacharacter attempts to extend the matching as far as possible.

So technically, the matching is correct: the match mathches the given pattern. However, it is likely the intent of the code is to match \<title> and <\title> seperately instead of the entire string.


___
2. **Lazy Matching**: Meaning match as little as possible.
___
It tries to take the shortest possible string that fits the pattern.

It matches as few characters as possible.

To make quantifiers lazy (non greedy), use **?** after the quantifer. This will match as few occurrence as possible.

| Quantifier Type | Greedy (default) | Lazy (minimal) | What it means                      |
| --------------- | ---------------- | -------------- | ---------------------------------- |
| Zero or more    | `*`              | `*?`           | Match as many / as few as possible |
| One or more     | `+`              | `+?`           | At least one occurrence            |
| Optional        | `?`              | `??`           | Zero or one occurrence             |
| Exact count     | `{m}`            | `{m}?`         | Exactly *m* times                  |
| Range           | `{m,n}`          | `{m,n}?`       | Between *m* and *n* times          |


Example, in 'aabbbb', 'ab*?' will match 'a'.

Is we want to match the openining snf closing tags separately, we can use non-greedy quantifiers.

Example:

```python
lazy = re.compile("<.*?>")

print(lazy.findall("<title>Regex<\title>"))
```

Now our findall has returned a list of two elements that we expexted - ['\<title>', '<\title>']

```python
#Last example
#Greedy
greedy_2 = re.compile(":.+:")

print(f"Greedy maching:\t{greedy_2.findall(":123: :123:")}")

#Lazy 
lazy_2 = re.compile(":.+?:")
print(f"Lazy matching:\t{lazy_2.findall(":123: :123:")}")
```

#### **Grouping and Capturing**
---
Grouping means wraping part of the regex in a parenthesis, which tells regex engine to treat what ever is inside as a single unit.

Grouping is achieved by enclosing a part of the regex pattern in parentheses()

* __Syntax: (pattern)__

Example, In the regex '(ab)+', the group (ab) is treated as a single unit followed by + quantifier. 

This matches one or more repetitions of 'ab' like, 'ababab', 'abab', etc.

**The Purpose of grouping**
___
1. Applying quantifiers to more than one character.
___
With grouping, quantifiers apply to more than one character only.

For example, "ab{3}" mathches "abbb" because "{3}" applies to "b" only.

(ab){3} matches "ababab" beacause {3} applies to the whole group.

This is the most basic and important use of grouping
___
2. Capture parts of matching - (Extraction).
___
When we use parenthesis, regex remembers what was matched - to be covered later

Example, "(\d{4})-(\d{2})-(\d{2})" applied on "2025-12-26" 

Captured groups:

* Group 1: '2025'
* Group 2: '12'
* Group 3: '26'

This is gold for data work.
___
3. Alternation(OR logic)
___
Grouping helps to say this or that.

For example: (cat|dog) matches cat or dog.

**Capturing groups**

---
By default, the groups capture and store part of the matched of the string by the group.

* Purpose: Capturing groups, store the part of the string matched by the group.

* Accessing captured groups: In python, captured groups can be accessed using the group() method of match objects.

Example: If a string "abca23" is matched against the regex '
(\\d+)', the part '123' is captured by the group and can be retrieved later

**group(0)** returns the entire match, while **group(1)** returns the first captured group.

code example:

```python

# The string to be searched
text = "abc123"
# Regex pattern with a capturing group for one or more 
# digits
pattern = r"(\d+)"
# Searching the text for the pattern
match = re.search(pattern, text)
# Checking if a match is found
if match:
    # Retrieving the first captured group
    captured_group = match.group(0)
    print("Captured Group:", captured_group)
else:
    print("No match found.")
```

**Non-capturing groups** 

---

* Purpose: Sometimes, we need the grouping functionality without capturing the matched string

* * __Syntax: (?:pattern)__

Example: The regex (?:ab)+ will match one or more repetitions of 'ab' but not store the match.

**Backreferences**

---

Backreference refers to something that was already matched by a capturing group.

In simple terms, regex matches something, stores it in a group and later says "match the same thing again"

Backreferences in a regex allow us to reuse the part of the string matched by a capturing group.

syntax: \n, where n is the group number.

Core idea:

( ... )   → capturing group
\1        → backreference to group 1
\2        → backreference to group 2

The number refers to the order of open parenthesis.

Example: The regex (\\d+)\\s\\1 matches two identical 
numbers separated by a space, like "42 42".

Example2: "(\w+)\s\1" matches "hello, hello", "data, data", "yes, yes", etc.

Explanation:

* (\w+) - Capture a word.
* Space.
* \1 - The same word again.

It will not match "Hello world"

**Named groups**

---

* Purpose: Improve the readability and manageability of complex regular expressions.

* * __syntax: (?p\<name>pattern)__

Example: (?P\<digits>\d+) creates a group named digits that matches one or more digits.

**Key points to note**

* **When you don't need the captured data, use non-captured groups to optimise performance.**

* **For complex patterns with multiple groups, use named groups for better readability.**

* If possible, avoid deeply reseted groups as they can make the regex pattern difficult to read and maintain.

#### Advanced anchors and boundaries
---

Anchors and boundaries are used to match positions within a string rather than specific characters.

They dont consume text, the just say "this spot must be here"

* __Word boundary( \b):__ 

Matches the position between a word character \w and a non-word character \W or the start/end of a string.

Use cases: To find words in text, ensuring that the match is not just part of a longer word.

Example pattern: \bword\b matches 'word' in 'word is a word', but not in 'swordfish'

Example 2: \bcat\b matches "cat", "cat!", "cat." not "scatter" or "concatenate"

* __Non-word boundary( \B):__ 

The opposite of \b, matches a position where there's no word boundary

Use cases: To match patterns that are part of a longer word.

Example pattern: \Bion\B matches 'ion' in 'ions are charged', but not in 'ion ia an atom'.

Example 2: \Bcat\B mathces "scatter", "concatenate"

* __Line anchor:^__

Matches the start of a string, or the start of a line if multiline mode is enable.

Use cases: To ensure that a pattern appears at the beginning of a text or a line.

Example: '^The' matches 'The start' but not in 'At the start'.

* __Line anchor:$__

Matches the end of a string or the end of a line if multiline mode is enabled.

Use cases: To ensure that a pattern appears at the end of a text or a line.

Example: 'end$' will match 'end' at the end of 'It's the end' but not in 'end of the story'.

---

**The Core Fact (this is the truth to remember)**

The caret ^ and the dollar sign are called anchors because they do not match characters, they match positions at the edges of a string.

What each one anchors.

* Caret ^: Anchors the match at the start of the string. eg "^Hello"

Means: "Hello" must appear at the very begining

* Dollar sign \$: Anchors the match at the end of the string. eg "word$"

Means: "Word" must appear at the very end

**Why they are called anchors**

Think of a string like a rope:

^--------------------$

* **^** anchors the left end.
* **$** anchors the right end.

The regex can't float around any more - it's fixed to a position.

**Key takeaway sentence**

Anchors restrict where a match can occur, not what is matched - When we use anchors, we are not telling regex what to match, we are telling it where the match is allowed.

That is the defining property.

**Why it matters**

Without anchors: \d{4} - Matches any where in the string.

With anchors ^\d{4}$ - Matches if the entire string is exactly four digits.

* **Anchors turn regex from searching into validation.**

* **One line summury- Caret and dollar are meta characters whose primary role is anchoring a regex to the start and end of a string**

It is a connection of the basics that we learnt at the begining, meta characters.

#### **re.search() function**
___
syntax:

* * **re.search(pattern, string, flags=0)**

The re.search() function is used to scan through a given string(string), looking for the first location where the regular expression pattern (pattern) matches.

How it works

1. Pattern matching: The function searches the string from the begining and stops as soon as it finds the first match for the pattern.

2. Flags: The flags argument is optional and can be used to modify certain aspects of the regex matching.

Return values:

* If a matching object is found, a match object containing information about the match is returned.

* If there's no match in the string, the function returns none.

Code example:

```python
text_2 = "Python is an amazing programing language. I wish people could know how amazing it is!"
pattern_2 = "amazing"

match_2 = re.search(pattern_2, text_2)

# print(match_2)
if match_2:
    print(f"Match found:\t{match_2.group()}")
    print(f"Position:\t{match_2.span()}")
    print(f"Start:\t{match_2.start()}")
    print(f"End:\t{match_2.end()}")
else:
    print("Match not found!")
```

#### **re.match() function**
---
syntax:

* * **re.match(pattern, string, flags=0)**

The re.match() function is used for finding matches at the begining of a string, unlike re.search(), which scans for the first occurence of the pattern across a whole string.

How it works:

1. Pattern matching: The function searches for a pattern match at the first character of a string.

2. Flags: The flags argument is optional. For example, re.MULTILINE or re.M makes the ^ and $ anchors match the start and end of each line, instead of just the start and end of the whole string.

Return values:

* If a match is found, a match object containing inforamtion about the match is returned

* If there's no match at the start of the string, the function returns None.

code example:

```python
pattern_3 = "Hello" 

text_3 = "Hello world!"

#Using re.match to find pattern at the start of the text
match_3 = re.match(pattern_3, text_3)

if match_3:
    print("match_3 found:", match_3.group())
    print("It is in positon:", match_3.start())
else:
    print("No match found")
```

#### **re.findall() function**
---
syntax:

* * **re.findall(pattern, string, flags=0)**

The re.findall() function searches a string for all non-overlaping occurrences of a regex pattern.

It's efficient for scanning large texts and extracting pieces of info that match a particular pattern.

How it works:

1. Pattern matching: It will scan the string from the left to the right, and for each pattern match, the matching strings will be added to a list.

2. Non-overlaping: Once a portion of the string is part of a match, it can't be part of another match.

Non-overlapping: matches "aa" at positions 0-1 and 2-3 → gives ['aa', 'aa']

Overlapping (if Python allowed it): could also match "aa" at positions 1-2 → would give ['aa', 'aa', 'aa']

3. Capturing groups: If the pattern includes capturing groups(parentheses), find all will return a list of groups. Multiple groups return a list of tuples.

Return values:

* A list of strings, where each string is a match of the pattern.

* If the pattern includes one or more capturing groups, it returns a list of groups(or tuples of groups).

Code example:

```python
text_4 = "The dates are 2023-01-01, 2024-02-02"

#Single capturing group
pattern_4 = r"(\d{4}-\d{2}-\d{2})"
#multiple capturing groups
pattern_multiple = r"(\d{4})-(\d{2})-(\d{2})"

dates = re.findall(pattern_4, text_4)
date_parts = re.findall(pattern_multiple, text_4)

print(dates)
print(date_parts)
```

#### **re.finditer() function**
---
Syntax:

* * **re.finditer(pattern, string, flags=0)**

re.finditer is used to find all non-overlapping occurrences of the pattern in the string just like re.findall, however, instead of returning a list of all the matches, it returns an iterator that yields match objects.

How it works:

1. Pattern matching: Scans the string from left to right; for each match, it yields a match object instead of the match string.

2. Iterative results: Returns an iterator yeilding match objects, allowing for iteration over each match in the string.

3. Non-overlaping: Similar to findall, once a portion of the string is part of a match, it won't be included in subsequent matches.

Return values:

* Each match object returned contains methods and atrributes that provide additional information about the match which include .group(), .start(), .end(), and .span().

Code example:

```python
text_5 = "The rain in Spain Stays mainly in the plain.Its Sites are very beautiful!"

pattern_5 = r"\bS\w+"

matches = re.finditer(pattern_5, text_5)

occurence=1
for match_ in matches:
    print("-"*50)
    
    print(f"Occurrence:{occurence}:\t{match_.group()}")
    print(f"Positions:\t{match_.span()}")
    print(f"Start:\t{match_.start()}")
    print(f"End\t{match_.end()}")
    occurence+=1

    print("-"*50)

# for match_ in matches:
#     print(f"Match:{match_.group()} at position: {match_.span()}, start:{match_.start()}, end:{match_.end()}")

```

#### **re.sub() function**
---
syntax:

* * **re.sub(pattern, rep1, string, count=0, flag=0)**

re.sub() is used to replace occurrences of a regex pattern in a string with a replacement string (rep1).

It can modify a string in various ways, from simple replacement to complex pattern based transformations.

How it works

1. Pattern identification: String is scanned for any match.

2. Replacement: Each match is repalced with rep1 string.

3. Count: if count is specifiedand greater than 0, only the first count occurences are replaced.
If count is 0(default), all the occurences are replaced.

Return value:

The method returns a new string with the specified substitutions made. The original string remains unchanged

Code example:

```python
text_6 = "I love python. Python is great for scripting. PytHon can be used fo data analysis."

pattern_6 = "Python"
replacement = "Java"

#Replace the first two occurrences of 'Python' with 'Java'
results = re.sub(pattern_6, replacement, text_6, count=2, flags=re.I) 

print(results)
print(text_6)
```

#### **re.split() function**
---
Syntax:

* * **re.split(pattern, string, maxsplit=0, flags=0)**

The re.split() method is used to split a string into a list using a specified regular expression pattern as the delimeter.

A delimeter is a special character or symbol used to seperate pieses of data in a tex or file.

How it works

1. Pattern identification: The re.split method scans the string and identifies all occurrences of the pattern.

2. Splittting: Whenever the pattern is found, the string is split at that location.

3. Maximum number of splits: If specified, the list will have at most maxsplit+1 elements. The default is '0', means "no limit".

Return value:

The method returns a list of strings obtained by spliting the input string at every match of the pattern.

Code example:

```python
text_7 = "Words, seperated, by, commas (and some spaces)"

pattern_7 = r",\s*" #Comma followed by zero or more spaces

results_2 = re.split(pattern_7, text_7)

print(results_2)
```

```python
#An example using compile
pat = re.compile(r",\s*")
rst = pat.split(text_7)
print(rst)
```

#### **re.compile() function**
---
Syntax:

* * **re.compile(pattern, flag=0)**

re.compile is used to compile a regex pattern into a regex object .

The primary reason for this method is performance optimisation, especially when the same regex is to be applied muiltiple times.

How it works:

1. Pattern compilation: The given pattern is compiled into an object that can be used to perform matches. 

This step involves parsing the pattern and converting it into an internal format optimised for matching.

2. Pattern identification: The regex engine uses this compiled pattern to efficiently identify matches in a string

Return values:

The method returns a regex object which can be used to perform varios regex operations like matching, search, find all, etc. 

This object stores the compiled version of the pattern, reducing the overhead of recompiling the regex each time it's used.

code example:

```python
#Compile the regex pattern for a simple word match
pattern_8 = re.compile(r"Hello")

#Simple text
text_8 = "Hello, world!"

#Using that compiled pattern to search in the text
match_4 = pattern_8.search(text_8)

if match_4:
    print("Match found")
else:
    print("No match found!")
```

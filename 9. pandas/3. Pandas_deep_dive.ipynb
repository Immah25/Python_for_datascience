#### __Accessing values withing a series__
---
To access values or elements within a series we use the __loc__ attribute, which stands fro location.

__loc__ allows us to access values or elements through there index label

We can also access a value or element in a series using the __iloc__ attribute which stands for integer location.

__iloc__ allows us to access a value using the integer location which is the index of the value

````python
#==================================================================================
#--------------------------------------using loc------------------------------------
#==================================================================================

#let's create a series that we will work with
#let's create Maths results for a certain class of seven students who have the following indexes
np.random.seed(7)
scores = np.random.randint(100, size=7)
print(scores)

maths_scores_se = pd.Series(scores, index = (101,102,103,104,105, 106,107))
print(maths_scores_se)

#Let's say we wanted to access the resuts of students with index:
#103
print(f"The student at index 103 scored {maths_scores_se.loc[103]}% in maths") #student at index 103

#105
print(f"The student at index 105 scored {maths_scores_se.loc[105]}% in maths") #marks for student at index 105

#=====================================================================================
#-----------------------------------using iloc----------------------------------------
#======================================================================================

# Index 0
print(f"The student at index 0 - 101 scored {maths_scores_se.iloc[0]}% in maths") #marks for student at index 0 - 101

# index 2
print(f"The student at index 2 - 103 scored {maths_scores_se.iloc[2]}% in maths") #marks for student at index 2 - 103
````

#### __Summury__
---

__1️⃣ Single Column vs Multiple Columns in `.iloc`__

* * Selecting a single column

```python
df.iloc[0, 1]  # first row, second column
```

* **No list needed** for a single column.
* Returns a **scalar value** (a single item).


* * Selecting a range of columns 

```python
df.iloc[0, 0:2]  # first row, columns 0 and 1
```

* You **don’t use a nested list** for a slice.
* `0:2` is a **slice object**, not a list.
* Returns a **Series** with the selected columns.

* * Selecting multiple specific columns (non-consecutive)

```python
df.iloc[0, [0, 2]]  # first row, columns 0 and 2
```

* Now you **do need a list** because these are discrete positions, not a continuous slice.
* Returns a **Series** with the selected columns.

---

✅ **Rule of Thumb**

| Selection Type                  | Syntax              | Use List? | Returns |
| ------------------------------- | ------------------- | --------- | ------- |
| Single column                   | `df.iloc[0, 1]`     | ❌         | Scalar  |
| Continuous range of columns     | `df.iloc[0, 0:3]`   | ❌         | Series  |
| Non-continuous/multiple columns | `df.iloc[0, [0,2]]` | ✅         | Series  |

**The same concept applies to loc as well!!!**

#### **Group by**
---

GroupBy is one of the most **powerful**, **important**, and **beautiful** concepts in Pandas — once you understand it, data analysis becomes 10× easier.

Let’s break it down **very simply**.

⭐ **What is `groupby()`?**

`groupby()` means:

* __1️⃣ Split the Data__

Divide your DataFrame into groups based on a column (or multiple columns).

* __2️⃣ Apply an operation__

For each group, run something like:

* sum
* mean
* count
* min
* max
* std (standard deviation)
* custom function

* __3️⃣ Combine the results__

Return a new DataFrame with the aggregated information.

Pandas does all 3 steps in the background.

📦 **Simple Example**

Say you have sales data:

| product | region  | sales |
| ------- | ------- | ----- |
| Phone   | Nairobi | 3000  |
| Tablet  | Mombasa | 2500  |
| Phone   | Kisumu  | 2800  |
| Tablet  | Nairobi | 2600  |

If you want **total sales per product**, you do:

```python
df.groupby("product")["sales"].sum()
```

Result:

| product | sales |
| ------- | ----- |
| Phone   | 5800  |
| Tablet  | 5100  |

🔥 **Why GroupBy Exists**

Because datasets usually contain:

* many repeated categories
* many repeated regions
* many repeated customers
* many repeated dates

GroupBy lets you **summarize** them.

🎯 **Most Common Uses**

* ➡️ Count how many times each value appears

```python
df.groupby("region")["sales"].count()
```

* ➡️ Average sales in each region

```python
df.groupby("region")["sales"].mean()
```

* ➡️ Summaries for multiple columns

```python
df.groupby("region")[["sales", "quantity"]].sum()
```

* ➡️ Group by multiple columns**

```python
df.groupby(["region", "product"])["sales"].sum()
```

This gives you a pivot-table-like output.

🧠 **Think of it like this:**

If your DataFrame was a classroom:

* `groupby("gender")` = group boys together and girls together
* apply mean → get average score for boys, average score for girls

🏁 **In Summary**

`groupby()` is for:

🔹 Splitting
🔹 Aggregating
🔹 Combining

It helps you turn messy, repetitive data into summarized insights.

We can perform group by on the whole dataframe or individual columns

#### **Group by aggregation - whole dataframe**

#### **String Standardiztion**
---

String standardization simply means **making text consistent** across a dataset so it is easier to clean, analyze, merge, or compare.

Think of it like “formatting all text to look the same.”

---

#### ✅ **In Data Cleaning (especially with Python), string standardization includes:**

#### **1. Lowercasing or uppercasing**

So that `"Kenya"`, `"kenya"`, and `"KENYA"` are treated the same.

```python
df["country"] = df["country"].str.lower()
```

---

#### **2. Removing extra spaces**

Cleaning `"  Nairobi "` → `"Nairobi"`.

```python
df["city"] = df["city"].str.strip()
```

---

#### **3. Removing multiple spaces between words**

`"Data    Science"` → `"Data Science"`

```python
df["text"] = df["text"].str.replace(r"\s+", " ", regex=True)
```

---

#### **4. Standardizing spelling variations**

* `"Mombasa"` vs `"Msa"`
* `"Nairobi"` vs `"Nrb"`
* `"Male"` vs `"M"`

We convert all of them into one version:

```python
df["city"] = df["city"].replace({
    "Msa": "Mombasa",
    "Mombas": "Mombasa",
    "Mombasa ": "Mombasa"
})
```

---

#### **5. Removing punctuation**

Cleaning names or product fields.

```python
df["name"] = df["name"].str.replace(r"[^\w\s]", "", regex=True)
```

---

#### **6. Replacing inconsistent formats**

Example: `"yes"`, `"Y"`, `"YES "`, `"y"` → all become `"yes"`.

---

#### **7. Converting to standardized patterns**

Example: phone numbers:
`"0792-752-985"` → `"0792752985"`

---

#### 📌 Why string standardization matters

Because datasets can’t be analyzed properly when:

* Nairobi ≠ nairobi
* Male ≠ M
* Ksh 1000 ≠ 1000
* “Product A ” ≠ “Product A” (because of trailing spaces)

It prevents:

* Wrong grouping
* Wrong counts
* Wrong merges
* Duplicate categories
* Failed joins in SQL/Pandas

---

#### 🎯 In short:

String standardization = **making all text consistent so your data becomes clean, uniform, and ready for analysis.**

### **Transforming data**
---

The `.apply()` method in **pandas** is one of the most powerful tools for **transforming or manipulating data** in a DataFrame or Series.

---

#### **What `.apply()` does**

`.apply()` lets you **run a custom function on each row or column** (or on each element if it’s a Series).

It’s like saying:

> “Hey pandas, take every value (or row/column) here and do this thing to it.”

---

#### **1️⃣ Using `.apply()` on a Series (column)**

Example: Convert scores to grades:

````python

data = {
    "name": ["Alice", "John", "Brenda", "David"],
    "score": [88, 75, 92, 60]
}

df = pd.DataFrame(data)

# Define a function
def grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    else:
        return "D"
    
df["grade"] = df["score"].apply(grade)
#We just pass the function as it is.
df
````

#### **2️⃣ Using `.apply()` with a lambda function**
---

You don’t always need to define a function separately:

````python
df["score_plus5"]=df["score"].apply(lambda x:x + 5)
df
````


#### **3️⃣ Using `.apply()` on a DataFrame (row-wise or column-wise)**
---
You can apply a function across **rows** or **columns**:

```python


# Sum along each row (axis=1)
df["total"] = df.apply(lambda row: row["age"] + row["score"], axis=1)
print(df)
```

* `axis=1` → apply function **row-wise**
* `axis=0` → apply function **column-wise**

---

#### **When to use `.apply()`**

* Converting scores to grades
* Transforming text (capitalizing, removing spaces)
* Creating new columns from existing ones
* Applying custom calculations on rows

---

#### ⚡ **Quick tip**

* If your operation is **element-wise**, `.apply()` works great.
* But for **large datasets**, **vectorized operations** (like `df["score"] + 5`) are faster than `.apply()`.

---

### When to use DataFrames
Unlike the NumPy array which is suited for storing and performing computations on homogeneous data (data of the same type), Pandas DataFrames can accommodate heterogeneous data. This makes them the choice data structure for manipulating often messy data (for example, tabular data from spreadsheets, or SQL tables). 

We should use a Pandas DataFrame if all of the following statements hold:

* We have 2-dimensional data (rows and columns).
* The data type is the same within a column.
* We are interested in the index (rows) and column names.

Pandas DataFrames are especially beneficial for data manipulation tasks like merging, joining, and reshaping data.

Important notes

When creating a new column from a series, it can be of any length but when creating a column from a list, it must be of the same length as the data frame.

creating a column from a single value:

dataframe["new_column"] = 4

or 

car_sales["passed_road_test"] = True
 
A new column will be created with fours, and True - bools respectively

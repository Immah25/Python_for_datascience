#### **Creating multiple sublots**

There are two options for creating subplots:

````python
# Subplot option 1

figure, ((axis1, axis2), (axis3, axis4))=plt.subplots(nrows=2,
                                                      ncols=2,
                                                      figsize=(15, 10));

#Let's plot some data
#Axis 1
axis1.plot(x, x/2);

axis1.set(
    title="Sample data",
    xlabel="x-axis",
    ylabel="y-axis"
)

#Axis 2
axis2.scatter(np.random.random(10), np.random.random(10));

axis2.set(
    title="Sample data",
    xlabel="x-axis",
    ylabel="y-axis"
)

#Axis 3
axis3.bar(nut_butter_prices.keys(), nut_butter_prices.values());

axis3.set(
    title="Kithusi's butter nut factory",
    xlabel="Type of butter nut",
    ylabel="Prices in $"
)

#Axis 4
axis4.hist(np.random.randn(1000));

axis4.set(
    title="Sample data",
    xlabel="x-axis",
    ylabel="y-axis"
);
````

````python
# Subplots option 2:

figure, axis = plt.subplots(nrows=2,
                            ncols=2, 
                            figsize=(15,10))

#Plot to each different index
#Axis 0 or axis index [0, 0]
axis[0, 0].plot(x, x/2)

axis[0, 0].set(
    title="Sample data",
    xlabel="x-axis",
    ylabel="y-axis"
)

# Axis 1 or index [0, 1]
axis[0,1].scatter(np.random.random(10), np.random.random(10))

axis[0,1].set(title = "Sample data",
         xlabel= "x-axis",
         ylabel= "y-axis")


#Axis[1,0]
axis[1,0].bar(nut_butter_prices.keys(), nut_butter_prices.values())

axis[1,0].set(
    title="Kithusi's nut butter factory",
    xlabel="x-axis",
    ylabel="y-axis"
)

# Axis[1,1]
axis[1,1].hist(np.random.randn(1000))

axis[1,1].set(title="Sample data",
              xlabel="x-axis",
              ylabel="y-axis");
````

#### Bins

Great question 😄 — and you’re *right* to pause here. This is one of those concepts that quietly makes or breaks how well you understand your data 📊✨

So… what are **bins**? 🧺

Think of a histogram as **grouping numbers into ranges**.

Those ranges are called **bins**.

Instead of plotting every single value, a histogram:

1. Splits the data range into **intervals (bins)**
2. Counts how many values fall into each interval
3. Draws bars whose height = count in that bin

Simple example 👇

Say your data is:

```text
[1, 2, 2, 3, 4, 5, 6, 7, 8, 9]
```

If you choose **3 bins**, pandas might create something like:

* **Bin 1:** 1 – 3 → count = 4
* **Bin 2:** 4 – 6 → count = 3
* **Bin 3:** 7 – 9 → count = 3

Each bar represents **how many values fall in that range**, not individual values.

Why bins matter ⚖️

* **Too few bins** → data looks oversimplified (you lose detail)
* **Too many bins** → chart looks noisy and misleading
* **Right number of bins** → you see the true shape of the data (skew, spread, outliers)

That’s why this is powerful:

```python
df.plot.hist(bins=30)
```

You’re telling pandas:

> “Show me the distribution with *this level of detail*.”

Why both methods exist 🤔

These two are actually the same under the hood:

```python
df.plot.hist(bins=30)
df.plot(kind="hist", bins=30)
```

The first one is just:

* clearer
* more readable
* harder to mess up

Big-picture intuition 🧠

> **Bins answer the question:**
> *“How is my data distributed across value ranges?”*

Once you get bins, histograms stop being mysterious and start becoming one of your best EDA tools 🔍🔥

#### Now let's look at how we can make some visualizations using the OO - Object-oriented method

What we have been using is ploting directly from pandas using the stateless method - the pyplot method

**Between pyplot and matplotlib OO method, which one should we use?**

- When plotting something quickly, its okey to use pyplot method - when we want to get a quick visualization
- When plotting something more advanced, use the OO method.

````python
# Let's say we wanted to explore patients who are over the age of 50
# We want to create a bit of visualization and do some data analysis

heart_disease_df.head()
````

````python
over_50 = heart_disease_df[heart_disease_df["age"] > 50]
over_50
````

````python
# pyplot method
# Let's create a scatter plot, with our x axis being age, y axis chol and color target
# Color is denoted with a c
over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target");

# The visualization is doing what its supposed to do but its not clear
````

````python
# The OO method - This is not purely the OO method but a combination of pyplot and OO method

figure, ax = plt.subplots(figsize=(10,6))
over_50.plot(kind="scatter",
             x="age",
             y="chol",
             c="target",
             ax=ax);

# ax.set_xlim([45, 100]);

# ax.set_xlim is going to change the limitation of our x axis to be from 45 to 100 instead of the default one we got, 50 to 80
# Matplotlip are good for creating axis limits that are suited to the data that we are using
````

````python
heart_disease_df
````

````python
# The OO method - pure -  from scratch

fig, ax = plt.subplots(figsize=(10, 6))

#Plot the data
scatter = ax.scatter(x = over_50["age"],
                     y=over_50["chol"],
                     c=over_50["target"]) #When we say color by target we simply mean, change the color of the dot deepending on the target value is 

#Customization
ax.set(title="Heart Disease and Cholesteral level",
       xlabel="age > 50",
       ylabel="cholestral");

#Let's add a legend
ax.legend(*scatter.legend_elements(), title="Target");

#The *scatter.legend_elements() method is going to grab the values from scatter - the varible we created above when ploting our data, unpack
#the legend elemnts which is going to query the c for color paarameter and look back at our heart_disease_df data frame to see what value are
# in the target column and use them to create the legend

# The object orientation method is better compared to the other methods

#Add a horizontal line

ax.axhline(over_50["chol"].mean(),
           linestyle="--");
````

````python
# Let's to create subplots
#Subplot of chol , age, thalach

figure, axis = plt.subplots(nrows=2,
                            ncols=1,
                            figsize=(15, 10),
                            sharex=True);

#Let's plot the data 

#----------------------------------------axis[0]-----------------------------------------------------
#Axis[0]
scatter1 = axis[0].scatter(x=over_50["age"],
                  y=over_50["chol"],
                  c=over_50["target"]);

# Customization
axis[0].set(title="Heart Disease and Cholestral",
            
            ylabel="Cholestral");

#Let's add a legend
axis[0].legend(*scatter1.legend_elements(), title="target");

#Let's add a hrizontal line
axis[0].axhline(over_50["chol"].mean(),
                linestyle="--")

#----------------------------------------axis[1]-----------------------------------------------------
#Axis[1]
scatter2=axis[1].scatter(x=over_50["age"],
                         y=over_50["thalach"],
                         c=over_50["target"]);

#Castomization
axis[1].set(title="Heart Disease and Thalech",
            xlabel="age > 50",
            ylabel="thalech")

#Let's add a legend
axis[1].legend(*scatter2.legend_elements(), title="target");

#Let's add a horizontal line
axis[1].axhline(over_50["thalach"].mean(),
                linestyle="--");

#Now let's add a title for the entire figure
figure.suptitle("Heat Disease Analysis", fontsize=16, fontweight="bold");
````

````python
# Customizing the legend, title, axis etc
#Set the style
plt.style.use("seaborn-v0_8-whitegrid")

# Create figure
figure, axis = plt.subplots(figsize = (10, 6))

#Plot or data
scatter5 = axis.scatter(x=over_50["age"],
             y=over_50["chol"],
             c=over_50["target"],
             cmap="winter") # This changes the color scheme

#Customization
axis.set(title= "Heart Disease and Cholesterol Levels",
         xlabel="age > 50",
         ylabel="Cholestral")

# add legend
axis.legend(*scatter5.legend_elements(), title="target")

# add a mean line
axis.axhline(over_50["chol"].mean(),
             linestyle="--");
````

````python

figure, (axis0, axis1)=plt.subplots(figsize=(15, 10),
                                    nrows=2,
                                    ncols=1, 
                                    sharex=True
                                    )

#----------------------------------------axis 0-----------------------------------------------------
#Axis0
#Ploting the data
scatter3 = axis0.scatter(x=over_50["age"],
              y=over_50["chol"],
              c=over_50["target"],
              cmap="winter")

#Customization
axis0.set(title="Heart Disease and Cholestral",
         
          ylabel="Cholestral")

#Change the x-limit
axis0.set_xlim([50, 80])

#Adding a legend
axis0.legend(*scatter3.legend_elements(), 
             title= "target");

#Lets add a horizontal line
axis0.axhline(over_50["chol"].mean(), 
              linestyle="--");

#----------------------------------------axis 1-----------------------------------------------------
# axis1
#plot the data
scatter4=axis1.scatter(x=over_50["age"],
                       y=over_50["thalach"],
                       c=over_50["target"],
                       cmap="winter")

#Customization
axis1.set(title="Heart Disease and Max Heart Rate",
          xlabel="Age > 50",
          ylabel="Max heart rate")

# Change the x limit
axis1.set_xlim([50, 80])

#Change the axis1 y limit
axis1.set_ylim([60, 200])

#Add a legend
axis1.legend(*scatter4.legend_elements(), 
             title="target")

#Add a mean horizontal line
axis1.axhline(over_50["thalach"].mean(), 
              linestyle="--");

#Let's add a title to the entire figure now
figure.suptitle("Heart Disease Analysis", fontsize=16, fontweight="bold");
````

#### **Introduction to Visualisations**

Data visualisation is the process of presenting data and information in graphical forms such as charts, maps, and diagrams. The goal of these graphics is to improve understanding of the underlying data by making patterns, trends, correlations, and outliers more visible than a simple study of the raw data.

There may be numerous occasions when we will need to share our data-driven results and insights. Unfortunately, raw data are not appropriate for this purpose. This is where maximising the potential of data visualisation becomes critical.

There are many different ways to represent the same data, and the choice of visualisation affects what information is conveyed to the audience.

Below is an image depicting some of the most popular charts used for data visualisation.

<center><img src="https://raw.github.com/Explore-AI/Pictures/master/Introduction_to_data_visualisation_edit.jpeg?raw=true" style="float: center; padding-bottom=0.5em" width=80%/></center>

We are going to look at what we call the four foundational visualization tools

1. **Bar graph or the column graph**

These are best used to show categorical data where each bar represents a category and the length or height of each bar the number of items in that category.

Ocationally, bar graphs can be used to show the change in a value over time, like temperatures across the months of the year but this is more often than not left for the line graph

2. **Pie/donut chart**

They are useful for showing the relative sizes or quantities contained in different categories like the difference in population sizes across 2 or 3 countries

3. **Line graphs** 

It is very popular for showing the changes of a particular value that occur in sequence for example over a period of time

Typically, time is shown on the x-axis, the horizontal one and  the height at each point in time is known as observation

4. **Scatter plots**

A scatter plot consist of a whole lot of points or markers and we can look for shapes and trends among the points to determine if there is a relationship between two varaibles

There can be any two numerical values, one ploted against the horizontal axis and the other against the vertical

So if we use the horizontal axis to show age and the vertical one to show income, then a person who is 35 years old and who earns 40000 dollars would result in a point where those two values intersect

Scatter plots are the primary visuals used to represent regression models and show corelation among variables

#### **Bar plot**

````python
import matplotlib.pyplot as plt
import numpy as np
````

* Create a dictionary with the number of donuts that each EDSA faculty ate.

* The dictionary named d**onuts_per_person** contains the following data

{"Andrew": 8, "Jaco": 7, "Jason": 5, "Jonathan": 8, "Kwanda": 6, "Laura": 6}

````python
donuts_per_person = {"Andrew": 8, "Jaco": 7, "Jason": 5, "Jonathan": 8, "Kwanda": 6, "Laura": 6}
donuts_per_person
````

````python
donuts = donuts_per_person.values()
people = donuts_per_person.keys()

y_pos = np.arange(len(donuts_per_person))


#Object oriented style

#Create a figure
figure, axis = plt.subplots()

#Plot the data

axis.bar(people, donuts, alpha=0.5); #The alpha key word is used the set the intensity or transparency of each of the bars 

#Customize
axis.set(title="How many donuts each",
         xlabel="People",
         ylabel="donuts");
````

````python
#Stateful/ pyplot style

plt.bar(y_pos, donuts, alpha=0.5)
plt.xticks(y_pos, people)
plt.ylabel("Donuts")
plt.title("How many donuts each")

plt.show()
````

#### **Analysis**

By looking at the graph, we can see that Andrew and Jonathan both at most donuts (8 each) and that jackson ate the fewest donuts (only 5).

---
#### **Pie Chart**

* Using the same dictionary, we will plot a pie chart
* Shows the propotionof the total number of donuts that each EDSA faculty member consumed (all of the donuts consumed)

````python
# The Object oriented method

# Prepare our data
sizes = donuts_per_person.values()
labels = donuts_per_person.keys()

# Create the figure and axis
fig, ax = plt.subplots()

#Ploting the pie chart
ax.pie(sizes, 
       labels=labels, 
       autopct="%1.1f%%",
       startangle=140,
       colors= ["gold", "yellowgreen", "lightcoral", "lightskyblue", "grey", "blue"]);

# What each part means:
# sizes → slice proportions
# labels → slice names
# autopct → show percentages
# startangle=90 → rotate for better orientation

# Making it a proper circle - Without this, our pie may look like an egg
ax.axis("equal")

#add tittle
ax.set_title("propotions of donuts ate by each person in EDSA", fontsize=10, fontweight="bold");
````

````python
# The pyplot or stateful method

#Data to plot
labels = donuts_per_person.keys()
sizes = donuts_per_person.values()
colors = ["gold", "yellowgreen", "lightcoral", "lightskyblue", "grey", "blue"]

# plotting the pie chart
plt.pie(sizes, 
        labels=labels, 
        colors=colors,
        autopct="%1.1f%%",
        startangle=140)

plt.axis("equal")

plt.title("Proportions of donuts ate by each person in the EDSA", fontsize=10, fontweight="bold");
````

#### **Analysis**

The pie chart shows us that Andrew and Jonathan ate the largest proportion(40% of the total) of the donuts!

````python
cols = ["Age", "Overall", "Acceleration", "Aggression", "Agility", "Balance", "Ball control",
        "Composure", "Crossing", "Curve", "Dribbling", "Finishing", "Free kick accuracy", "GK diving", 
        "GK handling", "GK kicking", "GK positioning", "GK reflexes", "Heading accuracy", "Interceptions",
         "Jumping", "Long passing", "Long shots", "Marking", "Penalties", "Positioning", "Reactions", 
          "Short passing", "Shot power", "Sliding tackle", "Strength", "Vision", "Volleys", "Sprint speed",
          "Stamina", "Standing tackle"]

soccer_df[cols] = soccer_df[cols].apply(pd.to_numeric, errors="coerce", axis=1)
````

Let's have a look at the descriptive statistics of each column to see if we can spot any anomalies.

````python
soccer_df.describe()
````

It would appear that we have `17981` entries in total, with some `NaN` values and some values that fall outside of the expected stats range.

We would have difficulty comparing an acceleration score of 67 with an anomaly score of 33270, where we expected all scores to fall below 100.

#### **Slicing**


Slicing is important for extracting a subset of a DataFrame. Using ``.iloc[]``, we will be able to slice the DataFrame similarly to the way we would slice a list. It uses number slicing to get **positional** records. The syntax for slicing is as follows:

```
dataframe.iloc[start_row:stop_row , start_column:stop_column]
``` 

When slicing in Pandas the start bound is included in the output, i.e. if we slice rows `[0:5]`, we will extract rows 0 to 4. 

Another slicing method is to use the ```.loc[]``` function native to Pandas. `.loc[]` allows us to slice based on data _labels_, as opposed to `.iloc[]` which slices based on the DataFrame _index_. An easy way to remember the difference is the **i** in `.iloc[]` is for **i**ndex.

````python
soccer_df.iloc[1:101,:] # selects rows 1 to 100 from the dataframe
````

````python
soccer_df.loc[1:3]
````

Pandas DataFrames are a great way to wrangle data, but to see trends and patterns, find key insights, and identify outliers in the data, it generally helps to visualise them in a meaningful way. 
#### **Histogram**
---



A histogram is very similar to a bar chart, except that it **groups** data or puts it into **ranges**. These groupings/ranges are called "**bins**".

```Seaborn``` and ```Matplotlib``` have their own histogram functions. 

```Seaborn``` will combine the histogram with a distribution if prompted to do so.
If we want to display the distribution, we simply add the argument ```kde=True``` to the plot.

Let's look at how to graph a histogram in both `Matplotlib` and `Seaborn` below.

````python
fig, ax = plt.subplots(nrows=2,
                       ncols=1,
                        figsize=(10, 5),
                        sharex=True)

sns.histplot(data=soccer_df, 
             x= "Age", 
             bins=5, 
             kde=True, 
             ax=ax[0], 
             alpha=0.5)

sns.histplot(data=soccer_df, 
             x="Age", 
             bins=5, 
             ax=ax[1], 
             alpha=1)

ax[0].set_title("Distribution plot with Histogram for ages",
                fontsize=10, 
                fontweight="bold")

ax[1].set_title("Histogram of age", 
                fontsize=10,
                fontweight="bold")

plt.tight_layout(); #Automatically adjust spacing so nothing overlaps - titles, axis labels, tick labels, multiple subplots.
````

````python
#We cannot use displot in seaborn because it creates its own histogram hence cannot combine with matplotlib
sns.displot(data=soccer_df, 
            x="Age", 
            bins=5, 
            kde=True);
````

````python
#Matplotlib version
fig, ax=plt.subplots()

ax.hist(soccer_df["Age"], bins=5)
ax.set_xlabel("Age");
````

**Note**

Seaborn automatically labels the x axis for you whereas for matplotlib, axis need to be specified !




**plt.tight_layout()**

What plt.tight_layout() means 🧠

In simple terms:

“Automatically adjust spacing so nothing overlaps.”

When you create plots with:

* titles
* axis labels
* tick labels
* multiple subplots

matplotlib doesn’t always space them nicely by default 😅

So labels can collide, get cut off, or overlap.

tight_layout() tells matplotlib to:

* inspect all axes

* calculate how much space each needs

* resize and reposition them cleanly

#### **Join plots**
---



```Seaborn```'s _jointplot_ will graphically display two variables on one plot and a single variable on the marginal sides of the graph. Essentially, a _jointplot_ allows us to see the relationship between two variables as well as each variable independently. 

Let's see an example below: 

````python
j = sns.jointplot(data=soccer_df, 
                  x="Overall", 
                  y="Age", 
                  kind="reg")

plt.suptitle("Joint plot showing the relationship between Overall Age of football players", 
             y=1.1,
             fontsize=10,
             fontweight="bold");
````

In the graph above, it is still difficult to see how the `Overall` and `Age` variables are related. To improve this, we can try changing the type of plot or _kind_ to a hexagon, which displays the number of matching points much more clearly. 

Looking at the code below, we added the argument ```kind='hex'```.

````python
j = sns.jointplot(data=soccer_df, 
                  x="Overall", 
                  y="Age", 
                  kind="hex")

plt.suptitle("Joint plot showing the relationship between Overall Age of football players", 
             y=1.1,
             fontsize=10,
             fontweight="bold");
````

#### **Box plots**
---



Box plots are used to show the spread of data. The main difference between a box plot and a histogram (which also shows a distribution as mentioned above) is that it takes up less space to plot. 

Box plots are constructed using the 5-number summary as indicated below:

```{Lower Extreme ; Q1 ; Median ; Q3 ; Upper Extreme}```

![Box plot](box_plot.png)


Let's see how the `Overall` stats compare based on different ages (from 23 to 26 years old):

````python
young_players = soccer_df[(soccer_df["Age"] > 22) & (soccer_df["Age"] <27)] #👉 Each condition must be wrapped in parentheses 
````

````python
plt.figure(figsize=(16,6))

sns.boxplot(data=young_players,
            x="Age",
            y="Overall");
````

From the plot above, we can see the general trends of `Overall` stats for each specific age. Box plots, compared to histograms, can make it much easier to compare groupings of data.

#### **Violin plot**
---



Violin plots are similar to box plots, but they use the KDE or distribution of values for groupings, in addition to a box (the box is still present within each graph). 

Let's plot one using ```Seaborn``` with the `Overall` stats for various `Age` categories (from 23 to 26 years old).

````python
plt.figure(figsize=(16, 6))

sns.violinplot(data=young_players,
               x="Age",
               y="Overall");
````

* The violin plots show us general distribution across each age, and it makes it much easier for us to compare them

#### **Facet grid**
---



Sometimes a useful approach (specifically when exploring medium-dimensional data) is to draw the same plot (a facet) on multiple subsets of the dataset, resulting in a matrix of visualisations (a facet grid). This technique is commonly referred to as “lattice”, or “trellis” plotting, and it is related to the idea of _small multiples_. 

Our data will need to be in a Pandas DataFrame for this to work.

Let's try this on our dataset.

````python
goalkeepers = soccer_df[(soccer_df['Preferred_Positions'].str.contains('GK'))]
````

````python
sns.pairplot(goalkeepers[['Age','GK diving','GK handling', 'GK kicking', 'GK positioning', 'GK reflexes']],
             hue='Age',palette="Spectral_r")
plt.show()
````
